### Interoperability & Zero-Copy Conversion

PyTorch and NumPy can share memory directly when running on CPU. This means converting between them takes **zero extra memory** and happens instantaneously.

```python
import numpy as np
import torch

# NumPy -> PyTorch (Shares memory buffer)
np_arr = np.array([1.0, 2.0, 3.0])
torch_tensor = torch.from_numpy(np_arr)

# PyTorch -> NumPy (Shares memory buffer)
torch_tensor = torch.tensor([1.0, 2.0, 3.0])
np_arr = torch_tensor.numpy()

```

#9.8 Math Reduction Operations
# PyTorch Tensor Reduction Operations

Reduction operations aggregate data along one or more dimensions, producing smaller tensors or scalar values.

| Category | Operation | PyTorch Function / Method | Example Code | Behavior / Notes |
| :--- | :--- | :--- | :--- | :--- |
| **Summation** | Total Sum | `torch.sum(A)` | `A.sum()` | Sum of all elements in tensor $A$. |
| | Row-wise Sum | `torch.sum(A, dim=1)` | `A.sum(dim=1)` | Collapses columns; sums across each row (shape $[M, N] \to [M]$). |
| | Column-wise Sum | `torch.sum(A, dim=0)` | `A.sum(dim=0)` | Collapses rows; sums across each column (shape $[M, N] \to [N]$). |
| | Keep Dimensions | `torch.sum(A, dim=0, keepdim=True)` | `A.sum(dim=0, keepdim=True)` | Keeps original rank (shape $[M, N] \to [1, N]$), useful for broadcasting. |
| **Averages & Products** | Mean (Average) | `torch.mean(A)` | `A.mean(dim=0)` | Requires floating-point or complex dtype. Computes $\frac{1}{N}\sum x_i$. |
| | Median | `torch.median(A)` | `A.median(dim=0)` | Returns the middle value (or value at index $\lfloor (N-1)/2 \rfloor$). |
| | Product | `torch.prod(A)` | `A.prod(dim=1)` | Computes product of all elements ($\prod x_i$). |
| **Statistics** | Standard Deviation | `torch.std(A)` | `A.std(dim=0, correction=1)` | Sample std dev (Bessel's correction $N-1$ default). Set `correction=0` for population. |
| | Variance | `torch.var(A)` | `A.var(dim=0, correction=1)` | Sample variance ($\sigma^2$). Set `correction=0` for population variance. |
| **Extrema & Indices** | Global Min / Max | `torch.min(A)` / `torch.max(A)` | `A.min()`, `A.max()` | Returns the single smallest or largest scalar value in tensor. |
| | Min / Max by Dim | `torch.min(A, dim=0)` / `torch.max(A, dim=0)` | `values, indices = A.max(dim=0)` | Returns a named tuple `(values, indices)` along specified dimension. |
| | Index of Min | `torch.argmin(A)` | `A.argmin(dim=1)` | Index of minimum value. Unrolled index if `dim` is omitted. |
| | Index of Max | `torch.argmax(A)` | `A.argmax(dim=1)` | Index of maximum value. Unrolled index if `dim` is omitted. |
| **Cumulative** | Cumulative Sum | `torch.cumsum(A, dim=1)` | `A.cumsum(dim=1)` | Returns cumulative sum along dimension (preserves tensor shape). |
| | Cumulative Product | `torch.cumprod(A, dim=0)` | `A.cumprod(dim=0)` | Returns cumulative product along dimension (preserves tensor shape). |

---

### Row / Column-wise Broadcasting Arithmetic

To perform row-wise or column-wise addition/subtraction between a 2D matrix $A$ (shape $[M, N]$) and a 1D tensor $B$:

* **Row-wise operation (apply $B$ to each row):**
  * Tensor $B$ must have shape $[N]$ or $[1, N]$.
  * Code: `A + B` or `A - B` (Broadcasting matches the trailing dimension automatically).
* **Column-wise operation (apply $B$ to each column):**
  * Tensor $B$ must have shape $[M, 1]$ (or unsqueezed explicitly).
  * Code: `A + B.unsqueeze(1)` or `A - B.unsqueeze(1)`.

#9.9 Math Matrix Operations

In PyTorch, **element-wise multiplication (`*`) is completely different from matrix multiplication (`@`)**.

### Matrix Multiplication (Dot Products)

| Operation | Operator | PyTorch Function | Behavior & Notes |
| :--- | :--- | :--- | :--- |
| **Vector Dot Product** | — | `torch.dot(v1, v2)` | Strictly for 1D tensors (vectors). Computes $\sum v_{1i} v_{2i}$. |
| **Matrix Multiplication** | `@` | `torch.matmul(A, B)` | Standard matrix multiplication. Supports broadcasting for higher dimensions. |
| **Strict 2D MatMul** | — | `torch.mm(A, B)` | Strictly for 2D matrices (no broadcasting). Faster if you know both are exactly 2D. |
| **Batch MatMul** | — | `torch.bmm(A, B)` | For 3D tensors: `[batch, n, m] @ [batch, m, p] -> [batch, n, p]`. |
| **Determinant** | — | `torch.det(A) or torch.linalg.det(A)` | For square 2D matrices or batches of square 2D matrices. |
| **Inverse** | — | `torch.inverse(A) or torch.linalg.inv(A)` | For 1D matrices. |

---

#### Matrix Inverse Formulas (for hand calculations)

##### 1. General Formula ($n \times n$)

$$A \cdot A^{-1} = A^{-1} \cdot A = I$$


$$A^{-1} = \frac{1}{\det(A)} \operatorname{adj}(A)$$

> **Requirement:** The matrix must be non-singular, meaning $\det(A) \neq 0$.

##### 2. Specific Formula ($2 \times 2$ Matrix)
Given a $2 \times 2$ matrix: $\det(A) = ad - bc$:

$$A^{-1} = \frac{1}{ad - bc} \begin{bmatrix} d & -b \\ -c & a \end{bmatrix}$$

### ⚠️ Common Pitfall: `reshape` vs. `transpose`

* **`A.transpose()`** mathematically alters the data layout (flips axes).
* **`A.reshape()`** keeps the same data sequence in memory but changes how it is grouped.
* **If you want a matrix transpose, NEVER use `.reshape()`. Use `.T`.**

## Dimension Discussioon
📐 Tensor Dimensions vs. Dataset Features

---

## 1. In PyTorch / Linear Algebra (Tensors)
In PyTorch, **dimension** refers to the **number of axes** (or `ndim`), not the size of the rows or columns.

* **0D Tensor (Scalar):** `torch.tensor(5)` — 0 axes. - just a number (5)
* **1D Tensor (Vector):** `torch.tensor([1, 2, 3])` — 1 axis (length 3).
* **2D Tensor (Matrix):** `torch.tensor([[1, 2], [3, 4]])` — 2 axes (rows and columns).
  * Axis 0 = Rows
  * Axis 1 = Columns
* **3D Tensor (e.g., Image Batch):** Shape `(32, 28, 28)` — 3 axes (Batch size, Height, Width).

---

## 2. In Data Science / Machine Learning (Datasets)
In a tabular dataset, **dimension = number of features (columns)**.

* **Rows ($N$):** Number of samples or data points.
* **Columns ($D$):** Number of features (dimensions).
* *Example:* A CSV file with 10,000 rows and 50 feature columns is a **50-dimensional dataset**.

---

| Context | What "Dimension" Means | Example |
| :--- | :--- | :--- |
| **PyTorch / NumPy** | Number of axes / shape length (`tensor.ndim`) | Shape `(2, 3)` = **2D Tensor** |
| **Data Science / ML** | Number of columns / features | 50 feature columns = **50-Dimensional Data** |

### `torch.softmax(input, dim)`

Softmax normalizes numbers so that they sum to `1.0`. The `dim` parameter specifies **which axis to normalize over**:

* `dim=1` (or `dim=-1`): Computes probabilities **per row** (across columns). Standard for multi-class predictions. **Sample's Class Probabilities**
* `dim=0`: Computes probabilities **per column** (across rows/batches). **not very useful in prediction**

````python
k = torch.tensor([10,20,30], dtype = torch.float32) # softmax requires float dtype
torch.softmax(k, dim=0) # dim 0 or -1, as no col or row for 1D tensor, only 1 single axis. so dim = 0, the first and only available axis. dim = 1 means second axis which is non existent
````

#9.13 Copying Tensor & Memory Discussion

##Shallow Copy:
- code: a = b
- shares the same memory. Same ID (memory address).
- Near Zero memory cost
- if u change one, the other will be changed too accordingly.
##Deep Copy or independent memory:
- code:
  - b = a.clone() (with preserving autograd history)
  - b = a.clone().detach() (standard for complete isolated copy)
  - b = a.copy.deepcopy() (also works in Python, pythonic way)

- will allocate new memory for b
- so will take the extra memory. Potentiality of memory waste.
- changes will be separate.

````python
a = torch.tensor( [1, 2, 3] )
print(f"The first a tensor: {a}")
b = a # shallow copy
# b = a.clone() # deep copy (still not standard)

# changing b
b [1] = 10
b

print(f"Tensor after changing b: {a}")
````
