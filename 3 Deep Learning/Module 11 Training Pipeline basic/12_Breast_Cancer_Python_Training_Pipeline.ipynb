## Defining the Model

```python
X_train_tensor.shape
```

```python
class MySimpleNN():

  def __init__(self, X): # constructor (self(to call w & b), input)
    # initializing weight (size: (30*1)) (to match the matmul operation without Transpose matrix)
    self.weights = torch.rand(X.shape[1], # num of rows = no of input/cols (num of col in input dataset)
                              1, # num of col = 1 # as all features going to only one neuron
                              dtype=torch.float64, # must be float, gradient of int type can't be calculated
                              requires_grad=True)  # keeps track

    # initializing bias (1*1)
    self.bias = torch.zeros(1, # 1 element matrix for the only one neuron
                            dtype=torch.float64,
                            requires_grad=True)

  def forward(self, X): # (self, input); Z = WX + b
    z = torch.matmul(X, self.weights) + self.bias
    y_pred = torch.sigmoid(z)
    return y_pred

  def loss_function(self, y_pred, y):

    # Clamp predictions to avoid log(0) or log(1) respective to undefined or 0.
    epsilon = 1e-7
    y_pred = torch.clamp(y_pred, epsilon, 1 - epsilon) # var = torch.clamp(var, min, max)

    # Calculate loss
    loss = -(y_train_tensor * torch.log(y_pred) + (1 - y_train_tensor) * torch.log(1 - y_pred)).mean()
    return loss
```

You can check the weights and biases after training the pipeline

### Important Parameters (lr and epoch)

```python
learning_rate = 0.1
epochs = 25
```

## Training Pipeline

```python
# create model
model = MySimpleNN(X_train_tensor) # training the model on train dataset

# define loop
for epoch in range(epochs):

  # forward pass
  y_pred = model.forward(X_train_tensor)

  # loss calculate
  loss = model.loss_function(y_pred, y_train_tensor)

  # backward pass
  loss.backward()

  # parameters update: so it renew every time ~ = the loop doesnt change the values each time of w_old.
  with torch.no_grad():
    model.weights -= learning_rate * model.weights.grad
    model.bias -= learning_rate * model.bias.grad

  # zero gradients
  model.weights.grad.zero_()
  model.bias.grad.zero_()

  # print loss in each epoch
  print(f'Epoch: {epoch + 1}, Loss: {loss.item()}') # epoch + 1 as count starts from zero
```

# 12.7 By inheriting nn module (Version 2)

We don't need to change anything from here as we are not changing the neural network, only making the code easier by integrating nn module (a feature in PyTorch).

### Importing nn Module

```python
import torch.nn as nn
```

## Defining the model (nn module)

```python
class MySimpleNN2 (nn.Module): # inheriting nn module in this class
  def __init__(self, num_features): # constructor # num_features = input_tensor[1] == num of inputs
    super().__init__() # nn module attribute involvement

    self.linear = nn.Linear (num_features, 1) # nn.Linear Transformation Operation (input nodes, output nodes) # 30 inputs and 1 direct output
    self.sigmoid = nn.Sigmoid () # the ACT was Sigmoid, so bringing that

  def forward (self, features):
    output = self.linear(features) # Linear Transformation Output
    # print(f"Linear Transformation Output: {output}")
    output = self.sigmoid(output) # it will save memory as being saved on the same storage
    # print(f"\nActivation Function Ouput: {output}")

  # do not need to write loss function def here, as nn has it's own BCELoss Function

    return output

```

Declaring learning rate and epochs for later.

```python
learning_rate = 0.1
epochs = 25
```

### Loss Function

```python
loss_function = nn.BCELoss()
```

```python
type(loss_function)
```

The `type(loss_function)` gives equation of the Loss Function.

## Training Pipeline

`model = model.double()`

double = float64, float = float32.

- class 1: MySimpleNN -> float64 inside code; to match that it was float64 in numpy array to tensor transformation.

- class 2: MySimpleNN2 -> Default NN Module has float32, so it ran into an error. That's why I had to add an extra line here, also I could change the code and make it float32 (Float) instead of float64 (Double).

- summary: it presents the report in float32 default.

####**Use float32 for the better**

```python
# creating the model
model = MySimpleNN2(X_train_tensor.shape[1]) # model = class_name (input) # input = col of X_train_tensor
model = model.double() # Cast model parameters to float64 to match input tensor

# optimizing
optimizer = torch.optim.SGD(model.parameters(), # SGD > GD (Sochastic Gradient Descent)
                            lr=learning_rate)  # lr = learning rate

# Defining the Loop
for epoch in range(epochs):

  # forward pass
  y_pred = model(X_train_tensor)

  # loss calculation
  loss = loss_function(y_pred, # y_pred.shape : 2D -> (batch_size, 1)
                       y_train_tensor.reshape(-1, 1)) # y_train : 1D -> (batch_size, )
                       # here reshape(-1,1): 1 is the no of col, -1 is saying to caldulate the necessary rows to store y_train in 1 col.
                       # As 2D can't be calculated with 1D, we are reshaping the y_train_tensor here.
  # ouput of loss: tensor(0.1748, dtype=torch.float64, grad_fn=<BinaryCrossEntropyBackward0>)

  # zero gradient : making everything zero before started backward passing , so yeah before backward pass
  optimizer.zero_grad()

  # backward : computes dL/dW and attaches them to weight.grad
  loss.backward() # why loss.backward? -> think about the computation graph (forward -> prediction -> loss -> backward)

  # updating parameters : wnre = wold - lr * dL/dw
  optimizer.step()

  # printing loss in each epoch
  print(f'Epoch: {epoch + 1}, Loss: {loss.item()}')
```
