## TensorDataset Example

(Pytorch already have given the DataSet Class)

```python
import torch
from torch.utils.data import TensorDataset, DataLoader
```

```python
# Dummy Dataset
features = torch.tensor([
    [1.0,2.0], # sample 1
     [3.0,4.0], # sample 2
      [5.0,6.0] # sample 3
    ])
labels = torch.tensor([
    0, # label for sample 1
    1, # label for sample 1
    0 # label for sample 3
    ])

# Making the dataset (sample, label)
dataset = TensorDataset(features, labels)
print(f"Integrating features and Labels together to make the dataset \n(2nd sample): {dataset[1]}\n") # integrating features and labels together

# dataloader
# you can change the num of batches and play urself
dataloader = DataLoader (dataset,
                         batch_size = 2, # 2 will be processed at a time
                         shuffle = True) # batches are randomized/shuffled

# Going through the dataset batch by batch
for batch in dataloader :
  features , label = batch # batch = features, label = batch[0], batch[1]
  print("feature :", features)
  print("label :", label)
```

```python
X,y = make_classification( # make_classification gives you (features, target)
    n_samples = 10, # total num of samaples
    n_features=2, # 2 features # shape of X = (10, 2)
    n_informative=2, # how many featues are incluencing the result? like t shirt size wont contibute to ur marks
    n_redundant=0, # duplicated or almost duplicated features
    n_classes=2, # two possible results (0, 1)
    random_state=42 # reproducibility
)
```

### Custom Dataset building the class

```python
from torch.utils.data import Dataset,DataLoader # creates CUstomDataset, feeds data to model in batches
```

```python
class CustomDataset(Dataset): # class creation
  def __init__( self,features,labels): # constructor
    self.features = features # Take the features that were given to me and store them inside this Dataset object under the name self.features
    self.labels = labels

  def __len__(self): # len(Dataset) will give u the len quickly ~= num of rows
    return self.features.shape[0]

  def __getitem__(self,index): # gives sample acc to index number
    return self.features[index], self.labels[index]
```

```python
dataset = CustomDataset(X,y) # creatin the dataset
# Dataset's features, labels = (X, y) # X, y has been defined earlier
```

## DataLoader

```python
dataloader = DataLoader(dataset,
                        batch_size=4,
                        shuffle=False)

# dataloader uses dataset.__len__() & dataset.__getitem___() to access the samples
```

```python
for batch_feature, batch_label in dataloader: # go through the dataloader one batch (features, labels) at a time
  print(batch_feature) # 4, 4, 2
  print(batch_label)
```
