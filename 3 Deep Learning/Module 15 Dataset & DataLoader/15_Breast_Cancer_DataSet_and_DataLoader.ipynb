````python
from torch.utils.data import Dataset,DataLoader
````

## Custom Dataset Class Creation
It is to make some useful functions which we can use for the rest of our code to call and use the dataset easily.

````python
class CustomDataset(Dataset): # class creation
  def __init__( self,features,labels): # constructor
    self.features = features # Takin the features that were given to me (self.) and store them inside this Dataset object under the name self.features
    self.labels = labels

  def __len__(self): # len(Dataset) will give u the len quickly ~= num of rows
    return self.features.shape[0]

  def __getitem__(self,index): # gives sample acc to index number
    return self.features[index], self.labels[index]
````

## Making the train & test dataset

````python
"""
new_dataset = CustomDataset( # calling the class
  X # feature tensor (must be tensor as the class is being called from torch)
  y # the label tensor
  )
"""

train_dataset = CustomDataset(X_train_tensor,y_train_tensor) # train_dataset = CustomDataset (features, labels)
test_dataset = CustomDataset(X_test_tensor,y_test_tensor)
````

# DataLoader

````python
train_loader = DataLoader(train_dataset,
                          batch_size=32,
                          shuffle=True)
test_loader = DataLoader(test_dataset,
                         batch_size=32,
                         shuffle=True) # doesnt need to be shuffled tho
````

# Defining the Model
1. 30 features goes to one neuron
2. Assign weights for each feature
3. Linear Transformation, Z = WX + b
4. Sigmoid, ouput = Sigmoid(Z)


````python
import torch.nn as nn
````

````python
# num of input features
X_train_tensor.shape[1]
````

````python
class MySimpleNN (nn.Module): # inheriting nn module in this class, requires train_features
  def __init__(self, num_features): # constructor # receives num of input features
    super().__init__() # nn.Module Class initialization # lets python tracks model's params

    self.linear = nn.Linear (num_features, 1)
    # nn.Linear Transformation Operation (input nodes, output nodes) # 30 inputs and 1 direct output
    self.sigmoid = nn.Sigmoid () # the ACT was Sigmoid, so bringing that

  def forward (self, features):
    output = self.linear(features) # Linear Transformation Output
    # print(f"Linear Transformation Output: {output}")
    output = self.sigmoid(output) # it will save memory as being saved on the same storage
    # print(f"\nActivation Function Ouput: {output}")

  # do not need to write loss function def here, as nn has it's own BCELoss Function

    return output
````

````python
learning_rate = 0.1
epochs = 25
````

## Parameters (need to feed the model X_train dataset to see the parameters)

````python
loss_function = nn.BCELoss()
````

# Training PipeLine

### Training Pipeline (with build DataLoader)

````python
#create model
model = MySimpleNN(X_train_tensor.shape[1])

#optimizer
optimizer = torch.optim.SGD (model.parameters(),
                            lr = learning_rate )

#define loop
for epoch in range(epochs):

  for batch_feature,batch_labels in train_loader:
    #forward pass
    y_pred = model(batch_feature)
    #Loss calculate
    loss = loss_function(y_pred,batch_labels.reshape(-1,1))

    # clear old gradient
    optimizer.zero_grad()
    # calculate new gradients
    loss.backward()
    # update weights and bias
    optimizer.step()



  #print loss in each epoch

  print(f" {epoch+1} , Loss = {loss.item()} ")
````

Note: Small technical note: this prints the loss from the last batch of each epoch, not the average loss over the whole epoch.

### Training pipeline with handmade dataloader

````python
#create model

model = MySimpleNN(X_train_tensor.shape[1])

#optimizer
optimizer = torch.optim.SGD( model.parameters(), lr = learning_rate )

#define loop
batch_size = 32
epochs = 25
n_sample = len(X_train_tensor)

for epoch in range(epochs): # repeat wtv underneath for 25 times

  # creating batches manually (unnecessary)
  # range(start, end, step) = (0, n_sample, batch_size) = start at 0, move forward by 32, do it till u reach the num of samples
  for start_indx in range(0,n_sample,batch_size):
    end_indx = start_indx + batch_size # calcualting current batch end

    # print(f" Start index {start_indx}, end indx {end_indx}\n")

    X_batch = X_train_tensor[start_indx:end_indx] # features
    y_batch = y_train_tensor[start_indx:end_indx] # labels

    #forward pass
    y_pred = model(X_batch) # y_pred = model(X_train)
    loss = loss_function(y_pred,y_batch.reshape(-1,1)) # (-1,1) as y_pred (x,) & y_batch(x, 1)


    #upate step

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()



  #print loss in each epoch

  print(f" {epoch+1} , Loss = {loss.item()} ")
````
