# Question 10 (10 Marks)
Handle missing values in the Titanic dataset by:
- Imputing missing Age values using SimpleImputer with mean strategy
- Imputing missing Embarked values using most frequent strategy
- Imputing missing Cabin values using constant strategy with "Missing"

Finally, check whether any missing values remain.


```python
# missing age value imputing with mean
# installing
from sklearn.impute import SimpleImputer
age_imputer = SimpleImputer(missing_values=np.nan ,strategy='mean') # np.nan er value gulo k mean diye impute korbe simpleImputer
age_imputer.fit(df_copy[['Age']]) # df r age col theke she mean ber kore np.nan (missing) col gulo khuje ber korbe (2D) - learn
df_copy['Age'] = age_imputer.transform(df_copy[['Age']]).ravel() # df r age col k transform krbe, 2D data k plain banabe 1D te ravel()
df_copy.sample(10)

# ~=
# age_imputer = SimpleImputer(strategy='mean') # variable declaration
# df['Age'] = age_imputer.fit_transform(df[['Age']]) # mean ber korbe, age col ta re lean krbe (fit) then mean diye (transform)

# missing embarked values using mode
embarked_imputer = SimpleImputer(missing_values=np.nan, strategy='most_frequent')
embarked_imputer.fit(df_copy[['Embarked']])
df_copy['Embarked'] = embarked_imputer.transform(df_copy[['Embarked']]).ravel()

# missing cabin values using "Missing" constant
cabin_imputer = SimpleImputer(missing_values=np.nan, strategy='constant', fill_value='Missing')
cabin_imputer.fit(df_copy[['Cabin']])
df_copy['Cabin'] = cabin_imputer.transform(df_copy[['Cabin']]).ravel()

# checking whether any missing values remain
df_copy.isnull().sum()
```
