The graph looks messy.

```python
# Main overall figure title (use suptitle instead of title)
plt.suptitle("Subplots in matplotlib")

plt.subplot(2, 2, 1)  # First subplot
plt.plot(x, y)
plt.title("The first subplot")

plt.subplot(2, 2, 2)  # Second subplot (fixing order)
plt.plot(x**2, x**2)
plt.title("The second subplot")

plt.subplot(2, 2, 3)  # Third subplot
plt.plot(y, x)
plt.title("The third subplot")

plt.subplot(2, 2, 4)  # Fourth subplot
plt.plot(x, y)
plt.title("The fourth subplot")

plt.tight_layout() # automatically adjust spacing to prevent overlapping

plt.show()
```

## Object Oriented Way & dpi
 - Object: Figure
 - dpi: dpi is the dots-per-inch (pixel per inch). The more the dpi, the more the resolution.

```python
fig = plt.figure()
```

The object (the canvas) of figure has been created. Now that we give the canvas a name, we can call the canvas and make changes to it.

`<Figure size 640x480 with 0 Axes>` = the height and width of the canvas, with no axes (no plot or subplots)

```python
fig = plt.figure(figsize = (20, 8),
                 dpi = 100) # in built 100
```

We have changed the canvas' width (20 = 2000 px) and height (8 = 800 px).

```python
fig = plt.figure(figsize = (20, 8))

# adding axes
axis1 = fig.add_axes([0.0, # axis distance from Y-axis
                      0.0, # axis distance from X-axis
                      0.2, # pixel (%) of width = 20% of 2000 px = 400 px
                      0.2]) # pixel (%) of height

axis2 = fig.add_axes([0.2, 0.2, 0.2, 0.2])
axis3 = fig.add_axes([0.4, 0.4, 0.2, 0.2])

plt.tight_layout()

plt.show()
```

Adding Axes

```python
fig = plt.figure(figsize = (20, 8))

fig.suptitle("Overall Figure Title")

# axis 1
axis1 = fig.add_axes([0.0, # axis distance from Y-axis
                      0.0, # axis distance from X-axis
                      0.2, # pixel (%) of width = 20% of 2000 px = 400 px
                      0.2]) # pixel (%) of height
plt.plot(x, y)
plt.title("The first axis of the figure")

# axis 2
axis2 = fig.add_axes([0.2, 0.2, 0.2, 0.2])
plt.plot(x**2, x**2)
plt.title("The second axis of the figure")

# axis 3
axis3 = fig.add_axes([0.4, 0.4, 0.2, 0.2])
plt.plot(y, x)
plt.title("The third axis of the figure")

# plt.tight_layout()

plt.show()
```

doesn't look good, title is not in the middle and too much gap between the title and the figures.

```python

fig = plt.figure(figsize=(20, 8))

# Position the main title directly over the top of your plot layout
fig.suptitle("Overall Figure Title",
             x=0.3, # midpoint of plot span 0.0 to 0.6.
             y=0.7) # lowers the title down so it sits cleanly above axis3. y - 0.2*3+0.1 = 0.7

# axis 1
axis1 = fig.add_axes([0.0, 0.0, 0.2, 0.2])
plt.plot(x, y)
plt.title("The first axis of the figure")

# axis 2
axis2 = fig.add_axes([0.2, 0.2, 0.2, 0.2])
plt.plot(x**2, x**2)
plt.title("The second axis of the figure")

# axis 3
axis3 = fig.add_axes([0.4, 0.4, 0.2, 0.2])
plt.plot(y, x)
plt.title("The third axis of the figure")

plt.show()
```

- width: num of cols in that ndarray

`array([1, 1, 1, 1])`:

- and alpha 1 (the last 1) means it's completely opaque, transparent.

## Designing

```python
import matplotlib.pyplot as plt
import numpy as np

# Sample Data
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)

# 1. FIGURE SETUP
plt.figure(
    figsize=(10, 5),  # width=10 inches, height=5 inches
    dpi=100           # Dots Per Inch (image resolution/sharpness)
)

# 2. PLOTTING & LINE STYLES
plt.plot(
    x, y1,
    label='Sine Wave',
    color='#1f77b4',     # Hex color code for the line
    linewidth=2.5,       # Thickness of the line in points
    linestyle='-'        # Line style: '-' (solid), '--' (dashed), '-.' (dash-dot), ':' (dotted)
)

plt.plot(
    x, y2,
    label='Cosine Wave',
    color='#ff7f0e',
    linewidth=2,
    linestyle='--'       # Dashed line style
)

# 3. TITLES, LABELS, AND PAD
plt.title(
    'Comprehensive Matplotlib Styling Example',
    fontsize=14,          # Size of the font
    fontweight='bold',    # Weight: 'normal', 'bold', 'heavy', 'light', 'semibold', 'medium'
    pad=15                # Distance (in points) between the title text and the top of the plot axes
)

plt.xlabel('X Axis (Time / Steps)', fontsize=11, fontweight='semibold')
plt.ylabel('Y Axis (Amplitude)', fontsize=11, fontweight='semibold')

# 4. AXIS LIMITS & XTICKS (USING NP.ARANGE)
plt.xlim(0, 10)          # Sets strict boundaries for X-axis: starts at 0, ends at 10 (hides data outside)
plt.ylim(-1.5, 1.5)      # Sets strict boundaries for Y-axis

# np.arange(start, stop, step) generates an array of numbers [0, 2, 4, 6, 8, 10].
# This forces tick marks to appear precisely at these values instead of letting Matplotlib auto-guess.
plt.xticks(np.arange(0, 11, 2))

# 5. LEGEND & FRAMEON
plt.legend(
    loc='upper right',    # Location options: 'upper left', 'upper right', 'lower left', 'lower right', 'center', 'best'
    frameon=True,         # (You asked about 'foramen' — this is 'frameon'): Toggles the legend background box (True/False)
    facecolor='#f9f9f9',  # Background color inside the legend box
    edgecolor='none',     # Border color of the legend box ('none' removes the border)
    fontsize=10
)

# 6. GRID & ALPHA
plt.grid(
    True,
    linestyle=':',        # Dotted grid lines
    alpha=0.6,            # Transparency level from 0.0 (invisible) to 1.0 (fully opaque)
    color='gray'
)

# 7. ANNOTATE (POINTING TO DATA)
plt.annotate(
    'Peak Sine',
    xy=(1.57, 1.0),       # The exact (x, y) coordinates of the point you want to point AT
    xytext=(2.5, 1.3),    # The (x, y) coordinates where the text label itself will sit
    arrowprops=dict(      # Dictionary controlling the pointer arrow styling
        facecolor='black',# Color of the arrow
        shrink=0.05,      # Space between arrow tip/base and the target text/point
        width=1,          # Thickness of the arrow stem
        headwidth=6       # Width of the arrow head
    ),
    fontsize=10,
    fontweight='bold'
)

# 8. LAYOUT & SAVING
plt.tight_layout()         # Automatically adjusts subplot parameters so labels/titles never get cut off
plt.savefig('styled_plot.png', dpi=300, bbox_inches='tight') # bbox_inches='tight' trims whitespace around borders
plt.show()
```
